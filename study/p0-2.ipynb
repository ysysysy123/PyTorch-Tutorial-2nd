```python
text = "1,10\n2,20\n3,30"

parts = text.split("\n")
sample=[]
for num in parts:
    values = num.split(",")
    x = int(values[0])
    y = int(values[1])
    sample.append({"x": x, "y": y})

print(sample)
print(sample[0])
print(sample[0]['x'])
```

```python
samples = []

with open("data.txt", "r", encoding="utf-8") as f:
    text = f.read()

for line in text.splitlines():
    values = line.split(",")

    x = int(values[0])
    y = int(values[1])

    samples.append({
        "x": x,
        "y": y
    })

print(samples)
```
