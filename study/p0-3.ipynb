Python 类（class）与对象（object）


````python
class Sample:
    def __init__(self, x, y):
        self.x = x
        self.y = y


sample = Sample(1, 10)

print(sample.x)
print(sample.y)

a = Sample(2, 20)
b = Sample(3, 30)

print(a.x)
print(b.y)
````

接下来理解 __init__ 和 self：

class Sample:
    def __init__(self, x, y):
        self.x = x
        self.y = y

当你写：

a = Sample(2, 20)

Python 会自动调用：

__init__(a, 2, 20)

这里的 self 就代表“当前正在创建的这个对象”，所以：

self.x = x

可以理解成：

把参数 x 保存到当前对象自己的 x 属性里

````python
class Sample:
    def __init__(self, x, y):
        self.x = x
        self.y = y

a = Sample(5, 50)

x = 100

print(a.x)
print(x)
````

接下来学类里的**方法 method**。看这个例子：

```python
class Sample:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def show(self):
        return self.x + self.y


a = Sample(2, 20)

print(a.show())
```

这里：

```python
def show(self):
```

表示这个函数属于 `Sample` 类，而：

```python
a.show()
```

会自动把 `a` 作为 `self` 传进去。

所以内部：

```python
self.x + self.y
```

实际上就是：

```python
a.x + a.y
```

也就是：

```python
2 + 20
```

````python
class Sample:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def double_x(self):
        return self.x * 2

a = Sample(3, 30)
b = Sample(5, 50)

print(a.double_x())
print(b.double_x())
````

````python
class MyDataset:
    def __init__(self, samples):
        self.samples = samples

    def size(self):
        return len(self.samples)

    def get(self, index):
        return self.samples[index]

samples = [
    {"x": 1, "y": 10},
    {"x": 2, "y": 20},
    {"x": 3, "y": 30},
]

dataset = MyDataset(samples)

print(dataset.size())
print(dataset.get(0))
print(dataset.get(2))
````

Python 的特殊方法（special / dunder methods）。

**Python 解释器会在特定语法 / 内置操作触发时自动调用**，用来重载类的行为。

Python 会自动对应为：

len(dataset)
    ↓
dataset.__len__()

dataset[1]
    ↓
dataset.__getitem__(1)


- `obj[key]` → 去找 `__getitem__`
- `obj[key] = val` → 去找 `__setitem__`
- `del obj[key]` → 去找 `__delitem__`
- `len(obj)` → 去找 `__len__`
- `obj + other` → 去找 `__add__`

````python
class MyDataset:
    def __init__(self, samples):
        self.samples = samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):
        return self.samples[index]

dataset = MyDataset(samples)

print(len(dataset))
print(dataset[0])
print(dataset[2]["y"])
````

````python
from torch.utils.data import Dataset, DataLoader

samples = [
    {"x": 1, "y": 10},
    {"x": 2, "y": 20},
    {"x": 3, "y": 30},
]

class MyDataset(Dataset):
    def __init__(self, samples):
        self.samples = samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):
        return self.samples[index]


dataset = MyDataset(samples)

loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)

for batch in loader:
    print(batch)
````
